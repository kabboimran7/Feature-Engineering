## Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_excel('supershops.xlsx')
df.head()
```

```python
label = LabelEncoder()
```

```python
df.Area = label.fit_transform(df.Area) #or label.fit_transform(df['Area']) 
```

```python
df.head()    #Alphabetical order 1st e Ctg er C=0, the Dhaka'r D=1 then Rangpur's R=2
```

```python
#Using Loops for more than one column handling of catagorical data
for column in df.columns:
    if df[column].dtype == np.number:  
        continue
    le = LabelEncoder()
    df[column] = le.fit_transform(df[column])
```

```python
df.head()
```

# One Hot Encoding  (Best Encoding Algorithm)

```python
#Step-01 : Creating dummy vars    ----- pd.get_dummies(df['Area'], dtype=int)
#Step-02 : Drop a column froms dummy columns  (parameetr = drop_first = True )        ----- pd.get_dummies(df['Area'], dtype=int, drop_first = True  )
#Step-03 : Droping that particular column from the DataFrame
#Step-04 : Concatinate modified dummy column and modifided dropped DataFrame
```

```python
df = pd.read_excel('supershops.xlsx')
df.head()
```

```python
dummy_variables = pd.get_dummies(df['Area'], dtype=int, drop_first = True )
```

```python
dummy_variables.head()
```

```python
new_df = df.drop('Area',axis=1)
```

```python
modified_df = pd.concat([new_df, dummy_variables],axis=1)  
```

```python
modified_df.head()
```

🧠 Dummy Variable Trap – বাংলা সংক্ষেপে ব্যাখ্যা:
Dummy Variable Trap হচ্ছে এমন একটি সমস্যা, যেখানে আমরা যখন একটি ক্যাটেগোরিকাল ভেরিয়েবলকে ডামি ভেরিয়েবল (0 আর 1) দিয়ে রিপ্রেজেন্ট করি, তখন যদি সবগুলো ডামি ভেরিয়েবল রাখি, তখন একটা ভেরিয়েবল অন্যগুলোর মাধ্যমে পুরোপুরি ভবিষ্যদ্বাণীযোগ্য হয়ে যায়।

⚠️ এর ফলে কী হয়?
মাল্টিকলি্নিয়ারিটি হয় (ভেরিয়েবলগুলোর মধ্যে সম্পর্ক তৈরি হয়),

রিগ্রেশন মডেল বিভ্রান্ত হয়,

কো-ইফিশিয়েন্ট (গুনাঙ্ক) সঠিকভাবে কাজ করে না।

✅ সমাধান:
একটা ক্যাটেগোরিকাল ভেরিয়েবলের জন্য যদি n টি ক্যাটেগরি থাকে, তাহলে শুধু n-1 টি ডামি ভেরিয়েবল রাখো।Python এ:

<b>pd.get_dummies(df['ColumnName'], drop_first=True)

এখানে drop_first=True দিলে একটা ক্যাটেগরি বাদ দেয়া হবে, তাই Dummy Variable Trap এ পড়বে না।

# OrdinalEncoder

```python
df = pd.read_excel('supershops.xlsx')   #Jodio ei dataset er jonne upojukto na
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
#Step-01 : Creating list of unique vals
#Step-02 : Creating an object of Ordinal class
#Step-03 : performing encode using Fit_trasform and 2D 
#Step-04 : Creating a DataFrame(column) of encoded values
#Step-05 : Replacing with previous column with encoded new column / DataFrame 
```

```python
df.Area.unique()
```

```python
#Step-01
city_list = ['Dhaka', 'Ctg', 'Rangpur']
```

```python
#Step-02 
ordinal = OrdinalEncoder(categories=[city_list])
```

```python
#Step-03 
encoded_value = ordinal.fit_transform(df[['Area']]) # number of sample and number of feature
```

```python
df.head()  # Encode er poro change hoy nai Area column so replace korbo
```

```python
#Step-04
new_area = pd.DataFrame(encoded_value, columns=['Area'])
new_area.head()
```

```python
#Step-05
df.Area = new_area
```

```python
df.head()  # We are now getting the changed / encoded area column
```
